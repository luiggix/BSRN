```python
path_origin=r"..\bsrn\fijo.dat"
path_destination="Prueba.dat"

day = [x.day for x in df["TIMESTAMP"]]
minute = [x.hour*60+x.minute for x in df["TIMESTAMP"]]
month = [x.month for x in df["TIMESTAMP"]]

with open(path_origin) as archive:
    lines = archive.readlines()    
    
lines.append("\n")
aux=True
for i in options.keys():
    if i == "C0100":
        lines.append("*"+i+"\n")
        print(i)

        dff=df[list(options[i])]
        dff.insert(0,"minute",minute)
        dff.insert(0,"day",day)

        dff.to_csv("nombre_archivo.txt", index=False, header=False)

        for j in range(len(dff[0:10])):
            cadena = dff[j:j+1].to_string(col_space=0,header=False,index=False,float_format='%.2f',na_rep="-999.99") + "\n"
            lines.append(cadena)
        print(cadena)
    
with open(path_destination, 'w') as archive:
    archive.writelines(lines)    
```
