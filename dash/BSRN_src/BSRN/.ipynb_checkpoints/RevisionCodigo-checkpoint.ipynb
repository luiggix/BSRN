```python
import pandas as pd
import numpy as np
import ipywidgets as widgets
import mysql.connector
```

```python
def get_credentials(usr, pswd, database, table, month, host = '132.248.182.174', port=3306):
    credentials = {
        'user': usr,
        'password': pswd,
        'host': host,
        'database': database,
        'port':port 
    }

    for k,v in credentials.items():
        print(k, v)

    # Convertimos la cadena de mes a número
    months = ['ene', 'feb', 'mar', 'abr', 'may', 'jun', 'jul', 'ago', 'sep', 'oct', 'dic']
    month = months.index(month) + 1
    print('Mes : ', month)
    # Obtenemos la cadena del año de la cadena de la base de datos
    db_year = database[-4:]

    print('file = ', database + '_' + months[month-1])
    
    try:
        conn = mysql.connector.connect(**credentials)
        
        if month != 1:
            query = "SELECT * FROM " + table + " where (month(timestamp)=" + str(month) + " or (month(timestamp)=" + str(month-1) + " and hour(timestamp)>17)" + ")" +" and year(timestamp)=" + db_year
        
        elif month == 1: # Tratamiento especial para el mes de enero
            db_pass_year = str(int(db_year) - 1)
            query = "select * from " + table + " where (month(timestamp)=12 and year(timestamp)="  + db_pass_year + " and day(timestamp)=31 and hour(timestamp)>17) or (month(timestamp)=1 and year(timestamp)=" +  db_year + ")"        

        print('Query : ', query)
        data = pd.read_sql(query, conn)
        data.to_csv(database + '_' + months[month-1])

        print(data)
        
        conn.close()
        
    except mysql.connector.Error as err:
        print('Error de conexión a la base de datos:', err)
```

```python
w_user = widgets.Text(
    value='',
    placeholder='Type something',
    description='Usuario',
    layout=widgets.Layout(width='190px'),
    disabled=False   
)

w_pswd = widgets.Text(
    value='',
    placeholder='Type something',
    description='Clave',
    layout=widgets.Layout(width='190px'),
    disabled=False   
)

w_db = widgets.Text(
    value='BSRN_2023', # Valor por omisión
    description='BD',
    layout=widgets.Layout(width='190px'),
    disabled=False   
)

w_table = widgets.Text(
    value='bsrn_2023_null', # Valor por omisión
    description='Tabla',
    layout=widgets.Layout(width='190px'),
    disabled=False   
)

w_month = widgets.Dropdown(
    options=['ene', 'feb', 'mar', 'abr', 'may', 'jun', 'jul', 'ago', 'sep', 'oct', 'dic'], 
    description='Mes',
    layout=widgets.Layout(width='190px'),
    disabled=False   
)

w_button = widgets.Button(
    description="", icon='play',
    layout=widgets.Layout(width='50px')    
)

ui = widgets.VBox([w_user, w_pswd, w_db, w_table, w_month, w_button])
ui.layout = widgets.Layout(border='solid 1px black')
ui.layout.width = '200px'

output = widgets.Output()

display(ui, output)

def on_button_clicked(b):
    output.clear_output(wait=True)
    with output:
        get_credentials(usr = w_user.value, 
                        pswd = w_pswd.value, 
                        database = w_db.value,
                        table = w_table.value,
                        month = w_month.value)

w_button.on_click(on_button_clicked)
```

```python
def data_processing(df, database, month):

    # Leemos el archivo correspondiente
    months = ['ene', 'feb', 'mar', 'abr', 'may', 'jun', 'jul', 'ago', 'sep', 'oct', 'dic']
    filename = database + '_' + months[month-1]
    df = pd.read_csv(filename, index_col=0)
    
    # Conversion de columna 'TIMESTAMP' al formato de fecha de pandas (datatime64)
    # Esto permite acceder a algunos parámetros de esta columna, como .dt.month 
    df['TIMESTAMP'] = pd.to_datetime(df['TIMESTAMP'], format='%Y-%m-%d %H:%M:%S')

    #df["TIMESTAMP"]=df["TIMESTAMP"] + pd.Timedelta(hours=6)

    # Eliminamos los renglones que no sean del mes que se está analizando
    df=df[df["TIMESTAMP"].dt.month == month]

    # Determinamos la fecha inicial y la fecha final
    start_date = df['TIMESTAMP'].iloc[0]
    end_date = df['TIMESTAMP'].iloc[-1]

#    print(start_date, end_date)

    # Conversion de la temperatura a Grados Kelvin
    to_kelvin = lambda t: t + 273.15
    df['CRPTemp_Avg']=to_kelvin(df['CRPTemp_Avg'])
    df['UVTEMP_Avg']=to_kelvin(df['UVTEMP_Avg'])
    df["DEW_POINT_Avg"] = to_kelvin(df["DEW_POINT_Avg"])

    # Agregamos la columna: GH_CALC_Avg - GLOBAL_Avg
    df["dif_GH_CALC_GLOBAL"] = df["GH_CALC_Avg"] - df["GLOBAL_Avg"]

    # Agregamos la columna: porcentaje de la diferencia
    df["percent_dif_GH_CALC_GLOBAL"] = 100 * abs(df["dif_GH_CALC_GLOBAL"]) / df["GH_CALC_Avg"]

    # Agregamos la columna: "dispersion" con los datos de df["GH_CALC_Avg"]
    df["dispersion"] = df["GH_CALC_Avg"]

    # Agregamos la columna datos de dispersion al 1%.
    df["percent_dispersion"] = 0.01 * df["dispersion"]

    # Agregamos una columna vacía
    df["empty"]=""

    # Agregamos una columna con valores enteros -999
    df["null_int"]=-999

    # Agregamos una columna con valores flotantes -99.9
    df["null_float1"]=-99.9

    # Agregamis una columna con valores flotantes -99.99
    df["null_float2"]=-99.99

    # Lista con todas las columnas del DataFrame
    columns = list(df.columns)

#    print(columns)

### <-------- NO VEO DONDE SE USAN ESTAS LISTAS -------->    
    # Listas con nombres de columnas sin su terminación (_Avg, _Std, _Min, _Mix)
    avg_list = [x.replace("_Avg","") for x in columns if x.endswith("Avg")]
    std_list = [x.replace("_Std","") for x in columns if x.endswith("Std")]
    min_list = [x.replace("_Min","") for x in columns if x.endswith("Min")]
    max_list = [x.replace("_Max","") for x in columns if x.endswith("Max")]
    
    # Determinamos los nombres comunes en las listas anteriores y los juntamos
    # en una lista
    common_list = list(set(avg_list) & set(std_list) & set(min_list) & set(max_list))

    # A esos nombres comunes les agregamos las terminaciones y los guardamos en una lista.
    aux = [elemento + "_" + str(i) for elemento in common_list for i in ["Avg","Std","Min","Max"]]

    # Eliminamos los nombres comunes de las columnas:
    unique_columns=columns.copy()    
    for i in aux:
        unique_columns.remove(i)

    unique_columns.remove("TIMESTAMP")
    unique_columns.remove("RECORD")
### <-------- --------------------------------- -------->    
    
    # Columnas del DataFrame organizadas por categoría.
    basic_parameters  = ["GLOBAL", "DIRECT", "DIFFUSE", "GH_CALC_Avg"]
    shortwave_balance = ["GLOBAL", "UPWARD_SW"]
    longwave_balance  = ["DOWNWARD", "UPWARD_LW", "DWIRTEMP", "UWIRTEMP", "CRPTemp_Avg"]
    meteorology       = ["CRPTemp_Avg", "RELATIVE_HUMIDITY_Avg", "PRESSURE_Avg", "DEW_POINT_Avg"]
    ultraviolet       = ["UVB", "UVTEMP_Avg", "UVSIGNAL_Avg"]
    statistics        = ["dif_GH_CALC_GLOBAL", "percent_dif_GH_CALC_GLOBAL", "dispersion"]

    categories = [basic_parameters, 
                  shortwave_balance, 
                  longwave_balance, 
                  meteorology, 
                  ultraviolet,
                  statistics]

    # others es un sinónimo de columns, cualquier acción sobre others afecta también a columns
#    others = columns 
#    for j in list(df.columns[-8:]):
#       others.remove(j)
#    categories.append(others)
    
    # El código anterior puede sustituirse por la siguiente línea:
    categories.append(columns[0:-8])
#    print(categories)

    names_categories=["Parámetros Básicos",
                      "Balance de onda corta",
                      "Balance de onda larga",
                      "Meteorología",
                      "Ultravioleta",
                      "Dispersión",
                      "Otros"]
    
    climatic_categories = dict(zip(names_categories, categories))

#    for k, v in climatic_categories.items():
#        print(k, v)
    
    return df
```
